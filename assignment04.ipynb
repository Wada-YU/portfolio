## **概要**

概要
評価したい「感じ（雰囲気や手応え）」を項目ごとに数値化し、現在のスコア・達成率（%）・伸ばせる点数（伸びしろ）を計算して結果を表示するシンプルなプログラムです。

複雑なクラスや設定を使わず、標準的な関数と辞書データ（dict）のみで構成されています。
1. どんな場面で使える？（具体的な活用事例）
① サービスの評価・改善（接客・店舗・Webサイト）
「デザインは好評だけど、使い勝手がいまいち」といった感覚的な評価を数値化して整理できます。

設定例：

「店舗の雰囲気」：90 / 100 pt

「接客の丁寧さ」：70 / 100 pt

「注文のしやすさ」：40 / 100 pt

効果：どこを改善すれば一番顧客満足度（イイ感じ度）が上がるのかが視覚的にわかります。

② プレゼン・企画案の「手応え」自己分析
上司やクライアントに見せる企画書の完成度や雰囲気を客観的に振り返る際に使えます。

設定例：

「ストーリー性」：85 / 100 pt

「デザインの見やすさ」：90 / 100 pt

「ワクワク感・インパクト」：50 / 100 pt

効果：感覚的に「何か足りない」と思っていた部分を「ワクワク感があと50pt伸ばせる（伸びしろ大）」と特定できます。

③ チームや自分の「モチベーション・コンディション」チェック
「最近なんかチームの雰囲気がいい／悪い」という抽象的なコンディションを数値化して共有するツールとして使えます。

設定例：

「コミュニケーションの取りやすさ」：80 / 100 pt

「作業の進めやすさ」：60 / 100 pt

「目標へのワクワク度」：30 / 100 pt
## **操作方法**

```python
def analyze_feeling(scores):
    """
    「感じ（雰囲気・印象・手応え）」をシンプルに分析する関数
    """
    results = []

    for item, score_info in scores.items():
        current = score_info['current']
        full = score_info['full']

        # 伸びしろ（あとどれくらい印象を伸ばせるか）
        margin = full - current
        # イイ感じ度（%）
        percentage = (current / full) * 100

        # 感覚・手応えに応じたステータス判定
        if percentage >= 90:
            status = "【完璧な感じ】"
        elif percentage >= 70:
            status = "【いい感じ】"
        elif percentage >= 50:
            status = "【まずまずな感じ】"
        else:
            status = "【伸びしろだらけ】"

        results.append({
            'item': item,
            'current': current,
            'full': full,
            'margin': margin,
            'percentage': percentage,
            'status': status
        })

    return results


def display_report(results):
    """
    分析結果を表示する関数
    """
    print("=" * 45)
    print(" ✨ フィーリング・印象の伸びしろ診断")
    print("=" * 45)

    total_current = 0
    total_full = 0

    for r in results:
        total_current += r['current']
        total_full += r['full']

        print(f"■ {r['item']}")
        print(f"  スコア: {r['current']} / {r['full']}pt (イイ感じ度: {r['percentage']:.1f}%)")
        print(f"  伸びしろ: あと +{r['margin']}pt 良くできます！ {r['status']}")
        print("-" * 45)

    total_margin = total_full - total_current
    overall_percentage = (total_current / total_full) * 100

    print("\n【全体まとめ】")
    print(f"  合計スコア: {total_current} / {total_full}pt")
    print(f"  全体の伸びしろ: あと +{total_margin}pt")
    print(f"  全体のイイ感じ度: {overall_percentage:.1f}%")
    print("=" * 45)


# --- テストデータ ---
sample_scores = {
    "デザインの雰囲気": {"current": 80, "full": 100},
    "接客の心地よさ": {"current": 65, "full": 100},
    "居心地のよさ": {"current": 40, "full": 100},
}

# 実行
analysis_results = analyze_feeling(sample_scores)
display_report(analysis_results)
```
